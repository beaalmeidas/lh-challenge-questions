```python
product_items = items_df[
    items_df["product_variant_id"].isin(
        product_variants_2["id"]
    )
].copy()
# product_items agora contem as ocorrencias de item em pedido que sao o produto estudado

print("Itens vendidos da variante:")
print(product_items.info())
print("\n")
print(product_items.head())
print("\n")
print(f"Quantidade de registros: {len(product_items)}")
print()
print("\n")
print(f"Unidades vendidas do produto 148: {product_items["quantity"].sum()}")
```

```python
print(
    f"Quantidade de pedidos: {product_items['order_id'].nunique()}"
)
```

```python
# pegando da tabela orders os pedidos com id em product_items
  # que possui os registros de venda do produto especifico
dataset = product_items.merge(
    orders_df[["id", "placed_at"]],
    left_on="order_id",
    right_on="id",
    how="inner"
)

print("Dataset unificado:")
print(dataset.info())
print("\n")
print(dataset.head())
print("\n")
print(f"Quantidade de registros: {len(dataset)}")
print()
```

```python
# transformando a coluna em data
dataset["placed_at"] = pd.to_datetime(
    dataset["placed_at"]
)


# criando coluna de mes no dataset
dataset["month"] = dataset["placed_at"].dt.to_period("M")

print("Dataset com mês:")
dataset.info()
print("\n")
print(
    dataset[
        ["order_id", "quantity", "placed_at", "month"]
    ].head()
)
print()
```

```python
# agrupando as vendas do produto por mes
monthly_sales = (
    dataset
    .groupby("month", as_index=False)["quantity"]
    .sum()
    .rename(
        columns={"quantity": "units_sold"}
    )
)

print("Vendas mensais:")
print(monthly_sales)
print()
```

```python
# alguns meses nao aparecem, por que teve 0 vendas do produto nele
  # exemplo: 2020-05, 2020-06
  # necessario marcar com 0 para nao inflar a contagem
```

```python
# adicionando meses que nao aparecem (pois tiveram 0 vendas do produto)
all_months = pd.period_range(
    start=monthly_sales["month"].min(),
    end="2026-03",
    freq="M"
)

monthly_sales = (
    monthly_sales
    .set_index("month")
    .reindex(all_months, fill_value=0)
    .rename_axis("month")
    .reset_index()
)

print("Série mensal completa:")
print(monthly_sales)
print()
```

```python
# separando dados de treino e teste
train = monthly_sales[
    monthly_sales["month"] <= pd.Period("2025-12")
].copy()

test = monthly_sales[
    monthly_sales["month"].between(
        pd.Period("2026-01"),
        pd.Period("2026-03")
    )
].copy()

print("Dados de treino:")
print(train)
print('\n')
train.info()
print('\n')

print("Dados de teste:")
print(test)
print()
test.info()
```

```python
# media movel dos ultimos 3 meses
predictions = []

history = train["units_sold"].tolist()

for _, row in test.iterrows():
    prediction = sum(history[-3:]) / 3
    predictions.append(prediction)
    last_3_months = history[-3:]
    print(
        f"Mês previsto: {row['month']} | "
        f"Últimos 3 meses: {last_3_months} | "
        f"Média: {prediction:.2f} | "
        f"Valor real: {row['units_sold']}"
    )
    history.append(row["units_sold"])

test["prediction"] = predictions


# previsoes pro 1o trimeste de 2026
```

```python
mae = mean_absolute_error(
    test["units_sold"],
    test["prediction"]
)

print(f"MAE: {mae:.2f} unidades")
```

```python
predictions = []

history = train["units_sold"].tolist()

for _, row in test.iterrows():
    prediction2 = sum(history[-6:]) / 6
    predictions.append(prediction2)
    last_6_months = history[-6:]
    print(
        f"Mês previsto: {row['month']} | "
        f"Últimos 6 meses: {last_6_months} | "
        f"Média: {prediction2:.2f} | "
        f"Valor real: {row['units_sold']}"
    )
    history.append(row["units_sold"])

test["prediction2"] = predictions


# previsoes utilizando ultimos 6 meses
```

```python
mae = mean_absolute_error(
    test["units_sold"],
    test["prediction2"]
)

print(f"MAE: {mae:.2f} unidades ()")
```

```python
predictions = []

history = train["units_sold"].tolist()

for _, row in test.iterrows():
    prediction3 = sum(history[-12:]) / 12
    predictions.append(prediction3)
    last_12_months = history[-12:]
    print(
        f"Mês previsto: {row['month']} | "
        f"Últimos 12 meses: {last_12_months} | "
        f"Média: {prediction3:.2f} | "
        f"Valor real: {row['units_sold']}"
    )
    history.append(row["units_sold"])

test["prediction3"] = predictions


# previsoes utilizando ultimos 12 meses

print("\n")
mae = mean_absolute_error(
    test["units_sold"],
    test["prediction3"]
)

print(f"MAE: {mae:.2f} unidades ()")
```

```python
# testando com o mesmo periodo, do ano anterior
sales_2025 = monthly_sales[
    monthly_sales["month"].dt.year == 2025
].copy()

test = monthly_sales[
    monthly_sales["month"].between(
        pd.Period("2026-01"),
        pd.Period("2026-03")
    )
].copy()

predictions4 = []

for _, row in test.iterrows():

    previous_year_month = row["month"] - 12

    previous_year_sales = monthly_sales[
        monthly_sales["month"] == previous_year_month
    ]["units_sold"]

    prediction = previous_year_sales.iloc[0]

    predictions4.append(prediction)

    print(
        f"Mês previsto: {row['month']} | "
        f"Mesmo mês de 2025: {previous_year_month} | "
        f"Previsão: {prediction} | "
        f"Valor real: {row['units_sold']}"
    )

test["prediction4"] = predictions4

print("\n")
mae = mean_absolute_error(
    test["units_sold"],
    test["prediction4"]
)

print(f"MAE: {mae:.2f} unidades")
```
