```python
import pandas as pd
import numpy as np
```

```python
orders_df = pd.read_csv("../1-lh_nautical_csv/orders.csv", encoding="utf-8")

orders_df.info()
```

```python
orders_df.head()
```

```python
orders_df['created_at'] = pd.to_datetime(orders_df['created_at'])

start_date = orders_df['created_at'].min()
end_date = orders_df['created_at'].max()
total_span = end_date - start_date

print(f"\nData inicial: {start_date}")
print(f"Data final: {end_date}")
print(f"Periodo de tempo: {total_span}")
```

```python
total_min_value = orders_df['total'].min()
total_max_value = orders_df['total'].max()
total_mean_value = orders_df['total'].mean()

print(f"\nValor minimo: {total_min_value}")
print(f"Valor maximo: {total_max_value}")
print(f"Valor medio: {total_mean_value:.2f}")
```
